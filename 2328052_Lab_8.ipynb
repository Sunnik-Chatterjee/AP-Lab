```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

1. Read Total profit of all months and show it using a line plot

```python
df = pd.read_csv('company_sales_data_matplotlib(1).csv')
plt.plot(df['month_number'],df['total_profit'])
plt.xlabel('Months')
plt.ylabel('Profits')
plt.title('Total Profits per month')
plt.show()
```

2. Get total profit of all months and show line plot with the following Style properties
Generated line plot must include following Style properties: –
Line Style dotted and Line-color should be red
Show legend at the lower right location.
X label name = Month Number
Y label name = Sold units number
Add a circle marker.
Line marker color as read
Line width should be

```python
plt.plot(
    df['month_number'],
    df['total_profit'],
    linestyle=':',
    color='red',
    marker='o',
    markerfacecolor='red',
    linewidth=3,
    label='Profit data of last year'
)
plt.xlabel('Month Number')
plt.ylabel('Sold units number')
plt.legend(loc='lower right')
plt.title('Total Profit per Month')
plt.show()
```

3. Read all product sales data and show it using a multiline plot
Display the number of units sold per month for each product using multiline plots. (i.e., Separate
Plotline for each product.

```python
plt.plot(df['month_number'], df['facecream'], label='Face Cream')
plt.plot(df['month_number'], df['facewash'], label='Face Wash')
plt.plot(df['month_number'], df['toothpaste'], label='Toothpaste')
plt.plot(df['month_number'], df['bathingsoap'], label='Bathing Soap')
plt.plot(df['month_number'], df['shampoo'], label='Shampoo')
plt.plot(df['month_number'], df['moisturizer'], label='Moisturizer')
plt.xlabel('Month Number')
plt.ylabel('Units Sold')
plt.title('Monthly Sales Data of Each Product')
plt.legend()
plt.show()
```

4. Read toothpaste sales data of each month and show it using a scatter plot

```python
plt.scatter(df['month_number'], df['toothpaste'], label='Toothpaste')
plt.xlabel('Month Number')
plt.ylabel('Units Sold')
plt.title('Toothpaste Sales per Month')
plt.legend()
plt.show()
```

5. Read face cream and facewash product sales data and show it using the bar chart

```python
x = np.arange(len(df['month_number']))
width = 0.35
plt.bar(x - width/2, df['facecream'], width, label='Face Cream')
plt.bar(x + width/2, df['facewash'], width, label='Face Wash')
plt.xlabel('Month Number')
plt.ylabel('Sales Units')
plt.title('Face Cream and Face Wash Sales')
plt.xticks(x, df['month_number'])
plt.legend()
plt.show()
```

6. Read sales data of bathing soap of all months and show it using a bar chart. Save this plot to your
hard disk

```python
plt.bar(df['month_number'], df['bathingsoap'], color='blue')
plt.xlabel('Month Number')
plt.ylabel('Sales Units')
plt.title('Bathing Soap Sales per Month')
plt.show()
plt.savefig('bathingsoap_sales.png')
```

7. Read the total profit of each month and show it using the histogram to see the most common profit
ranges

```python
plt.hist(df['total_profit'])
plt.xlabel('Profit Range')
plt.ylabel('Frequency')
plt.title('Total Profit Distribution per Month')
plt.show()
```

8. Calculate total sale data for last year for each product and show it using a Pie chart

```python
sales_data = [
    df['facecream'].sum(),
    df['facewash'].sum(),
    df['toothpaste'].sum(),
    df['bathingsoap'].sum(),
    df['shampoo'].sum(),
    df['moisturizer'].sum()
]
labels = [
    'Face Cream',
    'Face Wash',
    'Toothpaste',
    'Bathing Soap',
    'Shampoo',
    'Moisturizer'
]
plt.pie(
    sales_data,
    labels=labels,
    autopct='%1.1f%%',
    shadow=True
)
plt.title('Total Sales Contribution of Each Product (Last Year)')
plt.axis('equal')
plt.legend(fontsize=5, title="Products", title_fontsize=8)
plt.show()
```

9. Read Bathing soap facewash of all months and display it using the Subplot

```python
plt.subplot(1,2,1)
plt.plot(df['month_number'], df['bathingsoap'], color='blue')
plt.title('Bathing Soap Sales')

plt.subplot(1, 2, 2)
plt.plot(df['month_number'], df['facewash'], color='orange')
plt.title('Face Wash Sales')

plt.tight_layout()
plt.show()
```

10. Read all product sales data and show it using the stack plot

```python
plt.stackplot(
    df['month_number'],
    df['facecream'],
    df['facewash'],
    df['toothpaste'],
    df['bathingsoap'],
    df['shampoo'],
    df['moisturizer'],
    labels=['Face Cream','Face Wash','Toothpaste','Bathing Soap','Shampoo','Moisturizer']
)

plt.xlabel('Month Number')
plt.ylabel('Sales units')
plt.title('Stack Plot of Product Sales')
plt.legend(loc='upper left',fontsize=6,title='Products',title_fontsize=8)
plt.show()
```
